```python
import json
import os
import glob
import numpy as np
import matplotlib.pyplot as plt
```

```python
def calculate_average_accuracy_per_round():

    all_rounds_data = []

    for i in range(35):
        with open(f"/Users/georgexenos/Documents/flower/baselines/fedsim/results/fedsim_results_femnist/results_seed_{i}.json") as f:
            data = json.load(f)
            round_accuracies = [round_data['accuracy'] for round_data in data['round_res']]
            all_rounds_data.append(round_accuracies)

    all_rounds_data = np.array(all_rounds_data)

    average_accuracies = np.mean(all_rounds_data, axis=0)

    return average_accuracies.tolist()
```

```python
# Calculate and display the results
average_accuracies_sim = calculate_average_accuracy_per_round()
print(f"Number of rounds: {len(average_accuracies_sim)}")
print(f"Average accuracies per round:")
for i, acc in enumerate(average_accuracies_sim):
    print(f"Round {i+1}: {acc:.6f}")
```

```python
plt.plot(average_accuracies_sim)
```
